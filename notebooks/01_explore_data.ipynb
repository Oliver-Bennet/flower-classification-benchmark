```python
import os
from collections import Counter
from PIL import Image
import matplotlib.pyplot as plt

data_dir = r"D:\flower-classification-benchmark\data\flowers\train" 

classes = os.listdir(data_dir)
print("Số classes:", len(classes))
print("Classes:", classes)

# Đếm số ảnh mỗi class
counts = {cls: len(os.listdir(os.path.join(data_dir, cls))) for cls in classes}
print("Class distribution:", counts)

# Sample image
img_path = os.path.join(data_dir, classes[0], os.listdir(os.path.join(data_dir, classes[0]))[0])
img = Image.open(img_path)
print("Image size:", img.size)
plt.imshow(img)
plt.show()
```
